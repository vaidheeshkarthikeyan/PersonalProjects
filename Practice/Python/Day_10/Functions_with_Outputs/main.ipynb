# Functions with Outputs

```python
def format_name(f_name, l_name):
    name = f"{f_name} {l_name}"
    return name.title()
```

```python
print(format_name("vaidheesh", "karthikeyan"))
```

```python
def update_format_name(f_name, l_name):
    if f_name == "" or l_name == "":
        return
    name = f"{f_name} {l_name}"
    return name.title()
```

```python
print(update_format_name(f_name=input("Enter your first name:\n"), l_name=input("Enter your last name:\n")))
```
