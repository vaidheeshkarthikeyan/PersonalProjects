# Calculator

```python
from os import system
from art import ascii_art
```

```python
def add(number1, number2):
    return number1 + number2
```

```python
def subtract(number1, number2):
    return number1 - number2
```

```python
def multiply(number1, number2):
    return number1 * number2
```

```python
def divide(number1, number2):
    return number1 / number2
```

```python
operations = {
    "+": add,
    "-": subtract,
    "*": multiply,
    "/": divide
}
```

```python
def calculator(number1, number2, operation):
    result = operations[operation](number1, number2)
    print(f"{number1} {operation} {number2} = {result}")
    return result
```

```python
print(ascii_art)
continue_calculation = True

while continue_calculation:
    first_number = float(input("What's the first number?:\n"))
    for operation_key, operation_function in operations.items():
        print(operation_key)
    operation = input("Pick an operation:\n")
    second_number = float(input("What's the next number?:\n"))
    result_number = calculator(number1=first_number, operation=operation, number2=second_number)

    calculate_again = input(f"Type 'y' to continue calculating with {result_number}, or type 'n' to start a new calculation:\n")

    while calculate_again == 'y':
        system("cls")
        operation = input("+\n-\n*\n/\nPick an operation:\n")
        next_number = float(input("What's the next number?:\n"))
        result_number = calculator(number1=result_number, operation=operation, number2=next_number)
        calculate_again = input(f"Type 'y' to continue calculating with {result_number}, or type 'n' to start a new calculation:\n")

    continue_calculation = False
```
