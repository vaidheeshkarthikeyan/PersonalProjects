# Loops

## For loop

```python
fruits = ["Apple", "Banana", "Strawberry"]
```

```python
for fruit in fruits:
    print(fruit)
```

```python
for fruit in fruits:
    print(f"{fruit} pie")
```

```python
student_scores = [150, 142, 185, 120, 171, 184, 149, 24, 59, 68, 199, 78, 69, 85]
```

```python
sum = 0
for score in student_scores:
    sum += score

print(sum)
```

```python
max = 0

for score in student_scores:
    if score > max:
        max = score

print(max)
```

## For loops and the range() function

```python
for number in range(1, 10):
    print(number)
```

```python
sum = 0

for number in range(1, 101):
    sum += number

print(sum)
```
