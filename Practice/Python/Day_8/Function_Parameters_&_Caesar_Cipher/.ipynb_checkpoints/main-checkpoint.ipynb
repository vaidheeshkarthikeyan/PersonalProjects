# Functions with parameters

```python
def greet():
    print("Hi")
    print("Hello")
    print("Welcome!")
```

```python
greet()
```

```python
def greet(name):
    print(f"Hi {name}")
    print(f"Hello {name}")
    print(f"Welcome {name}")
```

```python
greet("Vaidheesh")
```

# Functions with multiple inputs

```python
def greet(name, location):
    print(f"Hi {name} from {location}! It was nice to meet you")
```

```python
greet("Vaidheesh", "Chennai")
```

```python
greet(name="Vaidheesh", location="Chennai")
```
