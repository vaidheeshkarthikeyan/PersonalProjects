# Caesar Cipher

```python
alphabets = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j',
             'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 
             'u', 'v', 'w', 'x', 'y', 'z']
```

```python
def encrypt(text, shift):
    
    encrypted_text = ""
    
    for letter in text:
        if letter in alphabets:
            new_index = (alphabets.index(letter) + shift) % 26
            new_letter = alphabets[new_index]
            encrypted_text += new_letter
        else:
            encrypted_text += letter
    print(encrypted_text)
```

```python
def decrypt(text, shift):

    decrypted_text = ""

    for letter in text:
        if letter in alphabets:
            new_index = (alphabets.index(letter) - shift) % 26
            new_letter = alphabets[new_index]
            decrypted_text += new_letter
        else:
            decrypted_text += letter
    print(decrypted_text)
```

```python
print(logo)

should_continue = True

while should_continue:
    direction = input("Type 'encode' to encrypt,  'decode' to decrypt:\n").lower()
    text = input("Type your message:\n").lower()
    shift = int(input("Type the shift amount:\n"))

    if direction == "encode":
        encrypt(text, shift)
    elif direction == "decode":
        decrypt(text, shift)
    else:
        print("Invalid input")
        break

    restart = input("Type 'yes' to go again, 'no' to end:\n").lower()
    if restart == "no":
        should_continue = False
```
