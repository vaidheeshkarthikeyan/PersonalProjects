# Number Guessing Game

```python
import random
import os
from ascii import ascii_art
```

```python
game_continues = True

while game_continues:
    random_number = random.randint(1, 101)
    os.system("cls")
    print(ascii_art)
    print(f"I'm thinking of a number between 1 and 100.")
    difficulty_choice = input("Choose a difficulty. Type 'easy' or 'hard':\n").lower()

    if difficulty_choice == 'easy':
        number_of_guesses = 10
    else:
        number_of_guesses = 5

    while number_of_guesses:
        guess_number = int(input("Make a guess:\n"))
        if guess_number == random_number:
            print(f"You got it! The answer was {guess_number}")
            if input("Do you want to play another game? Type 'y' for yes and 'n' for no:\n") == 'n':
                game_continues = False
            break
        elif guess_number > random_number:
            print("Too high.")
            number_of_guesses -= 1
            print(f"You have {number_of_guesses} chances left.")
        else:
            print("Too low.")
            number_of_guesses -= 1
            print(f"You have {number_of_guesses} chances left.")

        if number_of_guesses == 0:
            print("You've run out of guesses.")
            if input("Do you want to play another game? Type 'y' for yes and 'n' for no:\n") == 'n':
                game_continues = False
```
