# Scope

## Local Scope vs Global Scope

```python
enemies = 1

def increase_enemies():
    enemies = 2
    print(f"enemies inside function: {enemies}")
```

```python
increase_enemies()
print(f"enemies outside function: {enemies}")
```

```python
player_health = 10

def drink_potion():
    potion_strength = 2
    print(player_health)
```

```python
drink_potion()
```

### Does Python have Block Scope?

No, Python does not have Block Scope

```python
game_level = 3
enemies = ["Skeleton", "Zombie", "Alien"]

if game_level < 5:
    new_enemy = enemies[0]

print(new_enemy)
```

```python
def create_enemy():
    if game_level < 5:
        new_enemy = enemies[0]

    print(new_enemy)
```

```python
create_enemy()
```

## Modifying Global Scope

```python
enemies = 1

def increase_enemies():
    global enemies
    enemies += 1
    print(f"enemies inside function: {enemies}")
```

```python
increase_enemies()
print(f"enemies outside function: {enemies}")
```

```python
def increase_enemies(enemy):
    print(f"enemies inside function: {enemy}")
    return enemy + 1
```

```python
increase_enemies(enemies)
```
