```python
from os import system
```

```python
bid_dictionary = {}
```

```python
continue_bidding = True

while continue_bidding:
    name = input("What is your name?:\n").title()
    bid_amount = float(input("What is your bid?:\n$"))

    bid_dictionary[name] = bid_amount

    if input("Are there any other users who want to bid? Type 'y' for yes and 'n'\
             for no?\n") == 'n':
        continue_bidding = False
        
        max_amount = 0
        winner = ""
        for name, amount in bid_dictionary.items():
            if amount > max_amount:
                max_amount = amount
                winner = name

        print(f"Winner is {winner} with the bid of ${max_amount}")

        bid_dictionary = {}
    else:
        system("cls")
        continue_bidding = True
```
