# Tip Calculator

```python
print("Welcome to the tip calculator!")
total_bill = int(input("What was the total bill? $"))
tip = int(input("How much tip would you like to give? 10, 12, or 15? "))
person_split = int(input("How many people to split the bill?"))
tip_percentage = total_bill * (tip / 100)
total_amount = total_bill + tip_percentage
amount_each_person = total_amount / person_split
print(f"Each person should pay: ${round(amount_each_person, 2)}")
```
