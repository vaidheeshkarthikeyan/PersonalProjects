# Rock Paper Scissors

```python
import random
```

```python
rock_paper_scissors_list = ["rock", "paper", "scissors"]
```

```python
user_index = int(input("What do you choose? Type 0 for Rock, 1 for Paper or 2 for Scissors\n"))

user_choice = rock_paper_scissors_list[user_index]
computer_choice = random.choice(rock_paper_scissors_list)
print(f"User selects: {user_choice}")
print(f"Computer selects: {computer_choice}")

if user_choice == computer_choice:
    print("DRAW")
else:
    if user_choice == "rock":
        print("User WINS!" if computer_choice == "scissors" else "Computer WINS!")
    elif user_choice == "paper":
        print("User WINS!" if computer_choice == "rock" else "Computer WINS!")
    else:
        print("User WINS!" if computer_choice == "paper" else "Computer WINS!")
```
