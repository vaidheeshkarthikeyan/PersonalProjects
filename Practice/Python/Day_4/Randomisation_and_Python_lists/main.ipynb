# random module

```python
import random
```

```python
random_integer = random.randint(1, 10)
print(random_integer)
```

```python
random_float_number = random.random() * 10
print(random_float_number)
```

```python
random_floating_number = random.uniform(1, 10)
print(random_floating_number)
```

```python
random_number = random.randint(0, 2)

if random_number == 0:
    print("Heads")
else:
    print("Tails")
```

```python
print("Heads" if (random.randint(0, 2)== 0) else "Tails")
```

# Lists

```python
states_of_india = ["Tamil Nadu", "Kerala", "Karnataka", "Andhra Pradesh", "Maharashtra", "West Bengal", "NEw Delhi"]
```

```python
states_of_india[0]
```

```python
states_of_india[-3]
```

```python
additional_states = ["Punjab", "Haryana", "Himachal Pradesh", "Uttar Pradesh", "Uttarkhand"]
```

```python
friends = ["Alice", "Bob", "Charlie", "David", "Emanuel"]
```

```python
print(friends[random.randint(0, len(friends)-1)])
```

```python
print(random.choice(friends))
```
