# Functions

## Built-in functions

```python
print("Hello")
num_char = len("Hello")
print(num_char)
```

## User-defined functions

```python
def my_function():
    print("Hello")
    print("Bye")

my_function()
```

# While loop

```python
number = 10
i = 1

while i <= number:
    print(i)
    i += 1
```
