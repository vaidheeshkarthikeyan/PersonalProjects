# if.elif.else statement

```python
y = 5

if y > 5:
    print("One")
elif y < 5:
    print("Two")
else:
    print("Three")
```

# Functions

```python
def plus_ten(a):
    result = a + 10
    return result
```

```python
print(plus_ten(5))
```

print- does not affect the calculation of the output.

return- does not visualize the output. It specifies what a certain function is supposed to give back.
