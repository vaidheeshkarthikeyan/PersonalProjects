# Poetic Chatbot

```python
from google import genai
from google.genai import types
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
```

```python
for model in client.models.list():
    print(model.name)
```

```python
system_prompt = "You are a poetic chatbot."

sample_content = [
    ("When was Google founded?",
     "In the late '90s, a spark did ignite, Google emerged, a radiant light. By Larry and Sergey, in '98, it was born, a search engine new, on the web it was sworn."),
    ("Which country has the youngest president?",
     "Ah, the pursuit of youth in politics, a theme we explore. In Austria, Sebastian Kurz did implore, at the age of 31, his journey did begin, leading with vigor, in a world filled with din.")
]
```

```python
def chatbot(prompt):
    contents = []
    for question, answer in sample_content:
        contents.append({"role": "user", "parts": [{"text": question}]})
        contents.append({"role": "assistant", "parts": [{"text": answer}]})
    contents.append({"role": "user", "parts": [{"text": prompt}]})

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=contents,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt,
            temperature=1,
            max_output_tokens=1000)
    )

    return response.text.strip()
```

```python
prompt = "When was Apple the company found?"
print(chatbot(prompt))
```
