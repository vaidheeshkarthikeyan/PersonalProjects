# Converting text to lowercase

## using .lower() method

```python
sentence = "This is a Test Sentence"
```

```python
lower_sentence = sentence.lower()
print(lower_sentence)
```

```python
sentence_list = ["This is Sentence 1",
                 "This is Sentence 2",
                 "This is Sentence 3"]
```

```python
lowercased_sentence_list = [sentence.lower() for sentence in sentence_list]
print(lowercased_sentence_list)
```

# Removing Stopwords

## eg: "and", "of", "a","to"

## NLTK - Natural Language Toolkit

1. Predefined lists of stopwords
2. Tools for tokenizing text
3. Datasets for experiments

```python
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
```

```python
en_stopwords = stopwords.words("english")
```

```python
print(en_stopwords)
```

```python
sentence = "It was too far to go to the shop and he did not want her to walk"
```

```python
sentence_with_no_stopwords = " ".join([word for word in sentence.split() if word not in en_stopwords])
```

```python
print(sentence_with_no_stopwords)
```

```python
en_stopwords.append("It")
```

```python
sentence_with_no_stopwords = " ".join([word for word in sentence.split() if word not in en_stopwords])
```

```python
print(sentence_with_no_stopwords)
```

# Regular Expressions

A special way of writing patterns to search through text

The `r` character indicates that the text is a raw string

```python
import re
```

```python
folder_path = r"C:\desktop\notes"
print(folder_path)
```

```python
result = re.search("pattern", r"string to contain the pattern") # checks whether a specified patterns appears within a string
```

```python
print(result)
```

```python
string = r'sara was able to find the items rather quickly'

new_string = re.sub("sara", "sarah", string) # searches for a specific text within a string and replaces it with new content
print(new_string)
```

`^` - start of the string,<br>

`|` - works like a `or`

# Tokenization

Breaking text into smaller tokens.<br>It is easier for analysis

'punkt_tab': a table used to figure out where sentences begin and end

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
sentences = "Her cat's name is Luna. Her dog's name is Max"
```

```python
sent_tokenize(sentences)
```

```python
sentence_1 = "Her cat's name is Luna and her dog's name is Max"
```

```python
word_tokenize(sentence_1)
```

# Stemming

Words are reduced to their base form.<br>It reduces the number of unique words.

1. Reduces complexity
2. Keeps the dataset easier to manage

Drawbacks: Output is not always meaningful

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
connect_tokens = ["connecting", "connectivity", "connected", "connector", "connect"]
```

```python
for token in connect_tokens:
    print(token, ": ", ps.stem(token))
```

# Lemmatization

It reduces a word to a meaningful base form while preserving its intended meaning.<br>It uses a predefined dictionary to keep the context of the word

Drawback: we may still have more unique words in our dataset compared to stemming

```python
nltk.download("wordnet")
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lm = WordNetLemmatizer()
```

```python
for token in connect_tokens:
    print(token, ": ", lm.lemmatize(token))
```
