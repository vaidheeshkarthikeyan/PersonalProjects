## POS tagging

spaCy: An NLP library that comes with pretrained models capable of recognizing
parts of speech, named entities and other linguistic features

```python
import torch
import spacy
import pandas as pd
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = "emma woodhouse handsome clever and rich with a comfortable home and happy disposition seemed to unite some of the best blessings of existence and had lived nearly twentyone years in the world with very little to distress or vex her she was the youngest of the two daughters of a most affectionate indulgent father and had in consequence of her sisters marriage been mistress of his house from a very early period her mother had died too long ago for her to have more than an indistinct remembrance of her caresses and her place had been supplied by an excellent woman as governess who had fallen little short of a mother in affection sixteen years had miss taylor been in mr woodhouses family less as a governess than a friend very fond of both daughters but particularly of emma between them it was more the intimacy of sisters even before miss taylor had ceased to hold the nominal office of governess the mildness of her temper had hardly allowed her to impose any restraint and the shadow of authority being now long passed away they had been living together as friend and friend very mutually attached and emma doing just what she liked highly esteeming miss taylors judgment but directed chiefly by her own"
print(text)
```

A spacy document: an object that stored all the text along with all the linguistic 
information spacy generates

```python
spacy_doc = nlp(text)
```

```python
pos_df = pd.DataFrame(columns=["token", "pos_tag"])
```

```python
for token in spacy_doc:
    pos_df = pd.concat([pos_df,
                        pd.DataFrame.from_records([{"token": token.text, 
                                                    "pos_tag": token.pos_}])],
                       ignore_index=True)
```

```python
pos_df.head(10)
```

```python
pos_df_count = pos_df.groupby(["token", "pos_tag"]).size().reset_index(name="counts").sort_values(by="counts", ascending=False)
```

```python
pos_df_count.head(10)
```

```python
pos_df_pos_tag_count = pos_df_count.groupby(["pos_tag"])["token"].count().sort_values(ascending=False)
```

```python
pos_df_pos_tag_count.head(10)
```

```python
nouns = pos_df_count[pos_df_count["pos_tag"] == "NOUN"][:10]
print(nouns)
```

The Bag of words only counts words. It doesn't consider their order or how important a word might be compared to others

## TF-IDF (term frequency-inverse document frequency)

It measures how important a word is within a collection of texts. It does this by looking at two things

1. how often a word appears in a single document.
2. how common is it across all documents

TF- how many times a word appears in a single document

IDF - how many times a word appears in the entire collection of documents

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidfvec = TfidfVectorizer()
```

```python
tfidfvec_fit = tfidfvec.fit_transform(data)
```

# Topic Modeling

Topic Modeling helps us uncover the main themes that run through a collection of documents

1. LDA (Latent Dirichlet Allocation)
2. LSA

## Latent Dirichlet Allocation (LDA)

1. Prepare the text
2. Build the structures that the LDA expects
3. Train the model using gensim.

```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
import gensim
import gensim.corpora as corpora
```

```python
data = pd.read_csv("news_articles.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
articles = data["content"]
```

```python
print(articles)
```

```python
articles = articles.str.lower().apply(lambda x: re.sub(r"[^\w\s]", "", x))

en_stopwords = stopwords.words('english')
articles = articles.apply(lambda x: " ".join([word for word in x.split() if word not in en_stopwords]))

articles = articles.apply(lambda x: word_tokenize(x))

ps = PorterStemmer()
articles = articles.apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
print(articles)
```

```python
dictionary = corpora.Dictionary(articles)
print(dictionary)
```

```python
doc_term = [dictionary.doc2bow(text) for text in articles] 
# doc2bow() takes a single article, looks up each word in the dictionary and return a list of word IDs together with their counts
```

```python
print(doc_term)
```

```python
num_topics = 2
```

```python
lda_model = gensim.models.LdaModel(corpus=doc_term,
                                   id2word=dictionary,
                                   num_topics=num_topics)
```

```python
lda_model.print_topics(num_topics=num_topics, num_words=5)
```

## Latent Semantic Analysis

1. Distributional Hypothesis
    - States that words with similar meanings appear frequently together
2. Singular Value Decomposition (SVD)

SVD is a method of Dimensionality reduction. Similarities can be identified by:
- Clustering
- Similarity Scores

```python
from gensim.models import LsiModel
```

```python
lsamodel = LsiModel(doc_term, num_topics=num_topics, id2word=dictionary)
print(lsamodel.print_topics(num_topics=num_topics, num_words=5))
```

A coherence score tells us how meaningful the top words in a topic are when grouped together

```python
from gensim.models.coherencemodel import CoherenceModel
import matplotlib.pyplot as plt
```

```python
coherence_values = []
model_list = []

min_topics = 2
max_topics = 11

for num_topics_index in range(min_topics, max_topics+1):
    model = LsiModel(doc_term, num_topics=num_topics_index, id2word=dictionary, random_seed=0)
    model_list.append(model)
    coherence_model = CoherenceModel(model=model, texts=articles, dictionary=dictionary, coherence="c_v")
    # How often the top words of a topic actually appear together in the documents
    coherence_values.append(coherence_model.get_coherence())
```

```python
plt.plot(range(min_topics, max_topics+1), coherence_values)
plt.xlabel("Number of Topics")
plt.ylabel("Coherence score")
plt.legend(("coherence_values"), loc="best")
plt.show()
```

```python
final_n_topics = 3
lsamodel_f = LsiModel(doc_term, num_topics=final_n_topics, id2word=dictionary)
print(lsamodel_f.print_topics(num_topics=final_n_topics, num_words=5))
```

## Logistic Regression

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.DataFrame([("i love spending time with my friends and family", "positive"),
("that was the best meal i've ever had in my life", "positive"),
("i feel so grateful for everything i have in my life", "positive"),
("i received a promotion at work and i couldn't be happier", "positive"),
("watching a beautiful sunset always fills me with joy", "positive"),
("my partner surprised me with a thoughtful gift and it made my day", "positive"),
("i am so proud of my daughter for graduating with honors", "positive"),
("listening to my favorite music always puts me in a good mood", "positive"),
("i love the feeling of accomplishment after completing a challenging task", "positive"),
("i am excited to go on vacation next week", "positive"),
("i feel so overwhelmed with work and responsibilities", "negative"),
("the traffic during my commute is always so frustrating", "negative"),
("i received a parking ticket and it ruined my day", "negative"),
("i got into an argument with my partner and we're not speaking", "negative"),
("i have a headache and i feel terrible", "negative"),
("i received a rejection letter for the job i really wanted", "negative"),
("my car broke down and it's going to be expensive to fix", "negative"),
("i'm feeling sad because i miss my friends who live far away", "negative"),
("i'm frustrated because i can't seem to make progress on my project", "negative"),
("i'm disappointed because my team lost the game", "negative")],
                    columns=["text", "sentiment"])
```

```python
data = data.sample(frac=1).reset_index(drop=True)
```

```python
x = data["text"]
y = data["sentiment"]
```

```python
countvec = CountVectorizer()
```

```python
countvec_fit = countvec.fit_transform(x)
```

```python
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns=countvec.get_feature_names_out())
```

```python
print(bag_of_words)
```

```python
x_train, x_test, y_train, y_test = train_test_split(bag_of_words, y, test_size=0.3, random_state=7)
```

```python
lr_model = LogisticRegression(random_state=1).fit(x_train, y_train)
```

```python
y_pred = lr_model.predict(x_test)
```

```python
accuracy_score(y_pred, y_test)
```

```python
print(classification_report(y_pred, y_test, zero_division=0))
```
