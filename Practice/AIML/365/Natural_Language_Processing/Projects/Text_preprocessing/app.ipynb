# Text Preprocessing

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import re
import pandas as pd
```

```python
data = pd.read_csv("tripadvisor_hotel_reviews.csv")
```

```python
data.info()
```

```python
data.describe()
```

```python
data.head()
```

```python
data["Review"][0]
```

```python
# Lowercasing the text
data["review_lowercase"] = data["Review"].str.lower()
```

```python
data.head()
```

```python
# Removing Stopwords
en_stopwords = stopwords.words("english")
print(en_stopwords)
```

```python
en_stopwords.remove("not")
```

```python
data["no_stopwords_review"] = data["review_lowercase"].apply(lambda x: ' '.join([word for word in x.split() if word not in en_stopwords]))
```

```python
data.head()
```

```python
# Removing punctuations
data["review_with_no_punc"] = data["no_stopwords_review"].apply(lambda x: re.sub(r"[*]", "star", x))
```

```python
data.head()
```

```python
data["review_with_no_punc"] = data["review_with_no_punc"].apply(lambda x: re.sub(r"[^\w\s]", "", x))
```

```python
data.head()
```

```python
data["tokenized_review"] = data["review_with_no_punc"].apply(lambda x: word_tokenize(x))
```

```python
data.head()
```

```python
data["tokenized_review"][0]
```

```python
# Stemming reviews
ps = PorterStemmer()
```

```python
data["stemmed_review"] = data["tokenized_review"].apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
data.head()
```

```python
# Lemmatizing reviews
lm = WordNetLemmatizer()
```

```python
data["lemmatized_review"] = data["tokenized_review"].apply(lambda tokens: [lm.lemmatize(token) for token in tokens])
```

```python
data.head()
```

```python
data["lemmatized_review"][0]
```

```python
tokens_clean = sum(data["lemmatized_review"], [])
```

```python
unigrams = (pd.Series(nltk.ngrams(tokens_clean, 1)).value_counts())
```

```python
print(unigrams)
```

```python
bigrams = (pd.Series(nltk.ngrams(tokens_clean, 2)).value_counts())
```

```python
print(bigrams)
```
