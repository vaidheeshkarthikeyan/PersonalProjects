```python
import random
import os
from art import logo
```

```python
def deal_card():
    '''
    Returns a random card from the list of cards
    '''
    cards = [11, 2, 3, 4, 5, 6, 7, 8, 9, 10, 10, 10, 10]

    return random.choice(cards)
```

```python
def calculate_score(cards):
    '''
    Takes a list of cards and return the score of the list of cards
    '''
    if 11 in cards and 10 in cards and len(cards) == 2:
        return 0

    if 11 in cards and sum(cards) > 21:
        cards.remove(11)
        cards.append(1)
    
    return sum(cards)
```

```python
def compare(user_score, computer_score):

    if user_score == computer_score:
        return "DRAW!"

    if computer_score == 0:
        return "You LOSE! Computer has Blackjack!"
    if user_score == 0:
        return "You WIN with a Blackjack!"

    if user_score > 21:
        return "You went over! You LOSE! Computer WINS!"
    if computer_score > 21:
        return "Opponent went over! You WIN! Computer LOSES!"

    if user_score > computer_score: 
        return "You WIN! Computer LOSES!"
    if computer_score >user_score: 
        return "You LOSE! Computer WINS!"
```

```python
def blackjack():
    print(logo)
    user_cards = []
    computer_cards = []

    user_score = 0
    computer_score = 0

    game_over = False

    for _ in range(2):
        user_cards.append(deal_card())
        computer_cards.append(deal_card())

    while not game_over:
        user_score = calculate_score(user_cards)
        computer_score = calculate_score(computer_cards)

        print(f"Your Cards: {user_cards}. Current Score: {user_score}")
        print(f"Computer's first card: {computer_cards[0]}")

        if user_score == 0 or computer_score == 0 or user_score > 21:
            game_over = True
        else:
            user_card = input("Do you want to draw another card? Type 'y' for yes and 'n' for no:\n")
            if user_card == 'y':
                user_cards.append(deal_card())
            else:
                game_over = True

    while computer_score != 0 and computer_score < 17:
        computer_cards.append(deal_card())
        computer_score = calculate_score(computer_cards)

    print(f"Your Final hand: {user_cards}. Final Score: {user_score}")
    print(f"Computer's Final hand: {computer_cards}. Final Score: {computer_score}")
    print(compare(user_score, computer_score))
```

```python
game = input("Do you want to play a game of blackjack? Type 'y' for yes and 'n' for no:\n")
while game == 'y':
    os.system("cls")
    blackjack()
    game = input("Do you want to play a game of blackjack? Type 'y' for yes and 'n' for no:\n")
```
