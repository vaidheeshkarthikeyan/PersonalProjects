# POS Tagging

```python
nlp = spacy.load('en_core_web_sm')
```

```python
fake_news = data[data['fake_or_factual'] == 'Fake News']
```

```python
factual_news = data[data['fake_or_factual'] == 'Factual News']
```

```python
fake_spacydoc = list(nlp.pipe(fake_news['text']))
factual_spacydoc = list(nlp.pipe(factual_news['text']))
```

```python
def extract_token_tags(doc:spacy.tokens.doc.Doc):
    return [(i.text, i.ent_type_, i.pos_) for i in doc]
```

```python
fake_tags = []
columns = ['token', 'ner_tag', 'pos_tag']
```

```python
for index, doc in enumerate(fake_spacydoc):
    tags = extract_token_tags(doc)
    tags = pd.DataFrame(tags)
    tags.columns = columns
    fake_tags.append(tags)
```

```python
fake_tags = pd.concat(fake_tags)
```

```python
factual_tags = []
columns = ['token', 'ner_tag', 'pos_tag']
```

```python
for index, doc in enumerate(factual_spacydoc):
    tags = extract_token_tags(doc)
    tags = pd.DataFrame(tags)
    tags.columns = columns
    factual_tags.append(tags)
```

```python
factual_tags = pd.concat(factual_tags)
```

```python
fake_tags.head()
```

```python
fake_pos_count = fake_tags.groupby(['token', 'pos_tag']).size().reset_index(name='count').sort_values(by='count', ascending=False)
fake_pos_count.head(10)
```

```python
factual_pos_count = factual_tags.groupby(['token', 'pos_tag']).size().reset_index(name='count').sort_values(by='count', ascending=False)
factual_pos_count.head(10)
```

```python
fake_pos_count.groupby('pos_tag')['token'].count().sort_values(ascending=False).head(10)
```

```python
factual_pos_count.groupby('pos_tag')['token'].count().sort_values(ascending=False).head(10)
```

# Text Pre-Processing

```python
data.head()
```

```python
data['text_clean'] = data.apply(lambda x: re.sub(r"^[^-]*-\s", "", x["text"]), axis=1)
```

```python
data.head()
```

```python
data['text_clean'] = data['text_clean'].str.lower()
```

```python
data.head()
```

```python
data['text_clean'] = data.apply(lambda x: re.sub(r"[^\w\s]", "", x["text_clean"]), axis=1)
```

```python
data.head()
```

```python
en_stopwords = stopwords.words('english')
print(en_stopwords)
```

```python
data['text_clean'] = data['text_clean'].apply(lambda x: ' '.join([word for word in x.split() if word not in en_stopwords]))
```

```python
data.head()
```

```python
data['text_clean'] = data.apply(lambda x: word_tokenize(x['text_clean']), axis=1)
```

```python
data.head()
```

```python
lemmatizer = WordNetLemmatizer()
data['text_clean'] = data['text_clean'].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
```

```python
data.head()
```

```python
token_clean = sum(data['text_clean'], [])
```

```python
unigrams = (pd.Series(nltk.ngrams(token_clean, 1)).value_counts()).reset_index()[:10]
```

```python
print(unigrams)
```

```python
unigrams['token'] = unigrams['index'].apply(lambda x: x[0])

sns.barplot(
    x = 'count', 
    y = 'token',
    data = unigrams,
    orient = 'h', 
    palette = [default_plot_color],
    hue = 'token', 
    legend = False
).set(title='Most Common Unigrams after PreProcessing')
```

```python
bigrams = (pd.Series(nltk.ngrams(token_clean, 2)).value_counts()).reset_index()[:10]
```

```python
print(bigrams)
```

# Topic Modelling

```python
fake_news_text = data[data['fake_or_factual'] == 'Fake News']['text_clean'].reset_index(drop=True)
```

```python
dictionary_fake = corpora.Dictionary(fake_news_text)
```

```python
doc_term_fake = [dictionary_fake.doc2bow(text) for text in fake_news_text]
```

```python
coherence_values = []
model_list = []

min_topics = 2
max_topics = 11

for num_topics_i in range(min_topics, max_topics+1):
    model = gensim.models.LdaModel(doc_term_fake, num_topics=num_topics_i, id2word=dictionary_fake)
    model_list.append(model)
    coherence_model = CoherenceModel(model=model, texts=fake_news_text, dictionary=dictionary_fake, coherence='c_v')
    coherence_values.append(coherence_model.get_coherence())
```

```python
plt.plot(range(min_topics, max_topics+1), coherence_values)
plt.xlabel('Number of Topics')
plt.ylabel('Coherence Scores')
plt.legend(('coherence_values'), loc='best')
plt.show()
```

```python
num_topics_lda = 7
lda_model = gensim.models.LdaModel(corpus=doc_term_fake, id2word=dictionary_fake, num_topics=num_topics_lda)
```

```python
lda_model.print_topics(num_topics=num_topics_lda, num_words=10)
```

```python
def tfidf_corpus(doc_term_matrix):
    tfidf = TfidfModel(corpus=doc_term_matrix, normalize=True)
    corpus_tfidf = tfidf[doc_term_matrix]
    return corpus_tfidf
```

```python
def get_coherence_scores(corpus, dictionary, text, min_topics, max_topics):
    coherence_values = []
    model_list = []
    for num_topics_i in range(min_topics, max_topics+1):
        model = LsiModel(corpus, num_topics=num_topics_i, id2word=dictionary)
        model_list.append(model)
        coherence_model = CoherenceModel(model=model, texts=text, dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherence_model.get_coherence())

    plt.plot(range(min_topics, max_topics+1), coherence_values)
    plt.xlabel('Number of Topics')
    plt.ylabel('Coherence Score')
    plt.legend(('coherence_values'), loc='best')
    plt.show()
```

```python
corpus_tfidf_fake = tfidf_corpus(doc_term_fake)
```

```python
get_coherence_scores(corpus_tfidf_fake, dictionary_fake, fake_news_text, min_topics=2, max_topics=11)
```

```python
lsa_model = LsiModel(corpus_tfidf_fake, id2word=dictionary_fake, num_topics=7)
```

```python
lsa_model.print_topics()
```

# Creating our classification model

```python
data.head()
```

```python
x = [','.join(map(str, l)) for l in data['text_clean']]
y = data['fake_or_factual']
```

```python
countvec = CountVectorizer()
```

```python
countvec_fit = countvec.fit_transform(x)
```

```python
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns=countvec.get_feature_names_out())
```

```python
x_train, x_test, y_train, y_test = train_test_split(bag_of_words, y, test_size=0.3)
```

```python
lr = LogisticRegression(random_state=0).fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
accuracy_score(y_pred, y_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
svm = SGDClassifier().fit(x_train, y_train)
```

```python
y_pred = svm.predict(x_test)
```

```python
accuracy_score(y_pred, y_test)
```

```python
print(classification_report(y_test, y_pred))
```
