```python
# Importing the necessary packages

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re
import pandas as pd
```

```python
data = pd.read_csv('tripadvisor_hotel_reviews.csv')
data.info()
```

```python
data.head()
```

```python
data['Review'][0]
```

```python
# Convert text in the Review column to lowercase

review_lowercase = data['Review'].str.lower()
```

```python
data['Review_lowercase'] = review_lowercase
```

```python
data
```

```python
# Create our English Stopwords list.

en_stopwords = stopwords.words('english')
```

```python
en_stopwords.remove('not') #Decided to remove the 'not' word from our stopwords list
```

```python
en_stopwords
```

```python
# Extract words that are not present in stopwords

data['Review_no_stopwords'] = data['Review_lowercase'].apply(lambda x: ' '.join([word for word in x.split() if word not in en_stopwords]))
```

```python
data.head()
```

```python
data['Review_no_stopwords'][0]
```

# apply() function

It takes one columns and performs custom operation on every value in it

When we are preprocessing text, it is always worth making a new column for each of the steps in your preprocessing

```python
# Remmoving Punctuations using Regular Expressions

data['Review_no_punc'] = data.apply(lambda x: re.sub(r"[*]", " star", x["Review_no_stopwords"]), axis=1)
```

```python
data.head()
```

```python
data['Review_no_punc'] = data.apply(lambda x: re.sub(r"([^\w\s])", "", x["Review_no_punc"]), axis=1)
```

```python
data.head()
```

```python
# Tokenizing text using word_tokenize()

data['tokenized_review'] = data.apply(lambda x: word_tokenize(x['Review_no_punc']), axis=1)
```

```python
data.head()
```

```python
data['tokenized_review'][0]
```

```python
# Stemming using PorterStemmer

ps = PorterStemmer()
```

```python
data['stemmed_review'] = data['tokenized_review'].apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
data.head()
```

```python
#Lemmatizing the text using WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
data['lemmatized_review'] = data['tokenized_review'].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
```

```python
data['lemmatized_review'][0]
```

```python
# Preparing our text in right format
# Right now, each row in the 'lemmatized_token' column contains a seperate list of tokens
# Each review is stored as its own list of lemmatized words
# We need to combine these smaller lists into one longer list that contains every token from all reviews

tokens_clean = sum(data['lemmatized_review'], [])
```

```python
tokens_clean
```

```python
unigrams = pd.Series(nltk.ngrams(tokens_clean, 1)).value_counts()
```

```python
unigrams
```

```python
bigrams = pd.Series(nltk.ngrams(tokens_clean, 2)).value_counts()
```
