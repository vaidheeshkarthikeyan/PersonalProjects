```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import spacy
import re
import pandas as pd
import matplotlib.pyplot as plt
```

# Load Data

```python
bbc_news_data = pd.read_csv('bbc_news.csv')
```

```python
bbc_news_data.head()
```

```python
bbc_news_data.info()
```

```python
titles = pd.DataFrame(bbc_news_data['title'])
titles.head()
```

# Clean text data

### Lowercasing text

```python
titles['lowercase_title'] = titles['title'].str.lower()
titles.head()
```

### Removing Stopwords

```python
en_stopwords = stopwords.words('english')
titles['no_stopwords'] = titles['lowercase_title'].apply(lambda x: ' '.join([word for word in x.split() if word not in en_stopwords]))
titles.head()
```

### Removing punctuation

```python
titles['no_stopwords_no_punc'] = titles.apply(lambda x: re.sub(r"([^\w\s])", "", x['no_stopwords']), axis=1)
titles.head()
```

### Tokenize

```python
titles['tokens_raw'] = titles.apply(lambda x: word_tokenize(x['title']), axis=1)
titles['tokens_clean'] = titles.apply(lambda x: word_tokenize(x['no_stopwords_no_punc']), axis=1)
```

```python
titles.head()
```

### Lemmatizing

```python
lemmatizer = WordNetLemmatizer()
titles['tokens_clean_lemmatized'] = titles['tokens_clean'].apply(lambda x: [lemmatizer.lemmatize(token) for token in x])
titles.head()
```

```python
tokens_raw_list = sum(titles['tokens_raw'], [])
tokens_clean_list = sum(titles['tokens_clean_lemmatized'], [])
```

# POS Tagging

```python
nlp = spacy.load('en_core_web_sm')
```

```python
spacy_doc = nlp(' '.join(tokens_raw_list))
```

```python
pos_df = pd.DataFrame(columns=['token', 'pos_tag'])

for token in spacy_doc:
    pos_df = pd.concat([pos_df, pd.DataFrame.from_records([{'token': token.text, 'pos_tag': token.pos_}])], ignore_index=True)
```

```python
pos_df_counts = pos_df.groupby(['token', 'pos_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
pos_df_counts.head(10)
```

```python
nouns = pos_df_counts[pos_df_counts.pos_tag == "NOUN"][0:10]
nouns
```

```python
verbs = pos_df_counts[pos_df_counts.pos_tag == "VERB"][0:10]
verbs
```

```python
adjectives = pos_df_counts[pos_df_counts.pos_tag == "ADJ"][0:10]
adjectives
```

# NER

```python
ner_df = pd.DataFrame(columns=['token', 'ner_tag'])

for token in spacy_doc.ents:
    if pd.isna(token.label_) is False:
        ner_df = pd.concat([ner_df, pd.DataFrame.from_records([{'token': token.text, 'ner_tag': token.label_}])], ignore_index=True)
```

```python
ner_df.head()
```

```python
ner_df_counts = ner_df.groupby(['token', 'ner_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
ner_df_counts.head()
```
